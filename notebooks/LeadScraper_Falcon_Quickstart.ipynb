## 6. 🐍 Programmatic Python API
Use LeadScraper directly inside your Python applications or agent workflows with async/await support.

```python
from lead_scraper import LeadScraper

async with LeadScraper() as scraper:
    lead = await scraper.scrape_website("https://news.ycombinator.com")
    print(lead.display_card())
    print("Website:", lead.website)
    print("Emails:", lead.emails)
    print("Social Media Channels:", lead.social_media)
```

## 7. 📊 Inspect Auto-Exported Leads
Every run automatically exports structured `.json`, spreadsheet `.csv`, and formatted `.txt` cards into the `output/` directory.

```python
import glob
import pandas as pd

csv_files = sorted(glob.glob('output/*.csv'))
if csv_files:
    latest_csv = csv_files[-1]
    print(f"Loading: {latest_csv}")
    df = pd.read_csv(latest_csv)
    cols = [c for c in ['business_name', 'website', 'emails', 'phones', 'instagram', 'facebook', 'confidence_score'] if c in df.columns]
    display(df[cols])
else:
    print("Run an extraction above to generate leads.")
```
