## 3. 🌐 Direct Website URL Scrape
Extract contact information directly from a website URL and export to CSV.

```python
!lead-scraper-falcon --url "https://example.com" --output /content/example.csv
```

## 4. 🔍 Location Discovery Mode (Local Export)
Find businesses in a specific city/category and export leads to CSV without Google Sheets.

```python
!lead-scraper-falcon --discover "Dhaka" --quantity 3 --category "restaurant" --local --output /content/leads.csv
```

## 5. 📊 Inspect Extracted Leads with Pandas
Preview the enriched leads directly inside Colab.

```python
import pandas as pd
import os

if os.path.exists('/content/leads.csv'):
    df = pd.read_csv('/content/leads.csv')
    display(df[['business_name', 'phone_numbers', 'website', 'confidence_score']])
else:
    print('Run the discovery step above first to generate /content/leads.csv')
```
